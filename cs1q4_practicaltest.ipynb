```python
# Important Details
name = input("Enter student name: ")
while True:
    try:
        number_of_quiz_scores = int(input("Enter number of quiz scores to be inputted: "))
        if number_of_quiz_scores <= 0:
            print("Please enter a positive number of quiz scores.")
        else:
            break
    except ValueError:
        print("Invalid input. Please enter a whole number.")

# Quiz Scores
quiz_scores = []
for i in range(1, number_of_quiz_scores + 1):
    while True:
        try:
            score = float(input(f"Enter score {i}: "))
            quiz_scores.append(score)
            break
        except ValueError:
            print("Invalid input. Please enter a numerical score.")

# Total and Average Scores
def calculate_total(scores):
    total = sum(scores)
    return total

def calculate_average(scores):
    if len(scores) == 0:
        return 06
    average = sum(scores) / len(scores)
    return average

# Calculate and print results
total_score = calculate_total(quiz_scores)
average_score = calculate_average(quiz_scores)

print(f"\nStudent Name: {name}")
print(f"Total Score: {total_score}")
print(f"Average Score: {average_score}")

# Determine Pass or Fail
if average_score >= 60:
    print("Result: Pass")
else:
    print("Result: Fail")
```
